```python
%load_ext autoreload
%autoreload 2
import heapq               # findind the most frequent pairs
from pathlib import Path   # directory and file path manipulation
import urllib.request    # for downloading files from the web

import regex as re       # processing text with regular expressions

from tokenizer_functions import( download_token_data, initialize_vocab, count_tokens_in_file, enlarging_vocab)

intended_vocab_size = 10000

token_data_path = Path("data")                    # creating a Path object for the data directory

token_data_file_name="TinyStoriesV2-GPT4-train.txt"      


token_data_url = "https://huggingface.co/datasets/roneneldan/TinyStories/resolve/main/TinyStoriesV2-GPT4-train.txt"  #URL of the text file to download


file_path=token_data_path / token_data_file_name

PAT = r"""'(?:[sdmt]|ll|ve|re)| ?\p{L}+| ?\p{N}+| ?[^\s\p{L}\p{N}]+|\s+(?!\S)|\s+"""

dividing_str = "<|endoftext|>"

special_tokens=["<|endoftext|>"]



print("Stanford CS336 — Tokenizer")
print("This program will download a text file from the web, and then tokenize it into subword units.")
print("By Mehran Jalali, 2026")
```
